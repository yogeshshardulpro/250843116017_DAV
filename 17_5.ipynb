**Step 9: Detect Outliers using IOR(Select the final_marks colums for outliers analysis.)**

```python
# Select 'Final_marks'
final_marks = df['Final_marks']
# Calculate Q1 and Q3
Q1 = final_marks.quantile(0.25)
Q3 = final_marks.quantile(0.75)
# Calculate IQR
IQR = Q3 - Q1
# Calculate bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
# Display bounds
print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)
# Identify outliers
outliers = df[
    (df['Final_marks'] < lower_bound) |
    (df['Final_marks'] > upper_bound)
]
# Print outliers
print("\nOutliers:")
print(outliers[['student_Id', 'Final_marks']])
```

**Step 10: Identify Outliers.**

```python
outliers = df[
    (df["Final_marks"] < lower_bound) |
    (df["Final_marks"] > upper_bound)
]
# Print outliers
print("\nOutliers:")
print(outliers[["student_Id", "Final_marks"]])
# Print outlier count
print("\nNumber of Outliers:", len(outliers))
```
