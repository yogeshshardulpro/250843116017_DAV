**Practical 8 | Create interactive visualizations using Plotly Express and Plotly Graph Objects**

**Step 1 : Import Required Libraries**

```python
# import pandas, plotly express, plotly graph objects
import pandas as pd
import plotly.express as ps
import plotly.graph_objects as go
```

**Step 2 : Create Dataset**

```python
# Create Sales Records
data = {
    "Month" : ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Product" : ["Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet"],
    "Region" : ["North","South","East","West","North","South","East","West","North","South","East","West"],
    "Sales" : [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
    "Units_Sold" : [12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40],
    "Profit" : [18000, 27000, 30000, 22000, 32000, 35000, 28000, 34000, 38000, 33000, 46000, 50000]
}
```

```python
# Convert Sales Records into DataFrame
df = pd.DataFrame(data)
```

```python
# Display DataFrame
df
```

**Step 3 : Inspect the Dataset**

```python
# Display first five records
df.head()
```

```python
# Display the number of row and columns
df.shape
```

```python
# Display column names and data types
df.info()
```

**Step 4 : Create an Interactive Line Chart Using Plotly Express**

4.1 How does sales revenue change throughout the year

```python
fig = ps.line(
    df,
    x='Month',
    y='Sales',
    title='Sales vs Month'
)

fig.show()
```

**Step 5 : Improve the hover information**

```python
# Create a line chart using Sales over Month
# Add additional columns to the hover information. Students can see Product, Region, Units Sold
#  and Profit when they move the mouse over a data point
fig = ps.line(
    df,
    x='Month',
    y='Sales',
    title='Sales vs Month',
    hover_data=['Product','Region','Units_Sold', 'Profit']
)

fig.show()
```

**Step 6 :- Create an Interactive Bar Chart**

```python
# Group the dataset according to product
# Calculate the total sales generated by each product
grouped_by_product = df.groupby('Product')['Sales'].sum()
```

```python
# Create an interactive bar chart
fig = ps.bar(
    x=grouped_by_product.index,
    y=grouped_by_product.values,
    title='Total Sales by Product',
    labels={'x':'Product', 'y':'Total Sales'}
)

fig.show()
```

**Step 7 :- Interactive Scatter Plot**

```python
# To analyze the relationship between units sold and sales using an interactive scatter plot
fig = ps.scatter(
    x=df['Units_Sold'],
    y=df['Sales'],
    title='Units Sold vs Sales',
    labels={'x':'Units Sold', 'y':'Sales'}
)

fig.show()
```

**Step 8 :- Create an interactive Pie Chart**

```python
# To visualize regional sales contribution  using an interactive Pie Chart
fig = ps.pie(df, values='Sales', names='Region', title='Sales by Region')
fig.show()
```

**Step 9 :- Plotly Graph Objects**

```python
fig = go.Figure()

fig.add_trace(go.Line(
    x=df['Month'],
    y=df['Sales'],
    mode='lines+markers',
    name='Sales'
))

fig.show()
```
