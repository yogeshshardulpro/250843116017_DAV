**Step 1 : Import required Library**

```python
import pandas as pd
import matplotlib.pyplot as plt
```

**Step 2: Create a Dataset**

```python
data = {
    "Months" : ["January","February","March","April","May","June","July","August","September","October","November","December"],
    "Units_Sold" : [120,135,150,145,160,175,180,170,190,210,230,250],
    "Revenue" : [120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000, 250000],
    "Profit" : [20000, 23000, 25000, 24000, 27000, 30000, 31000, 29000,33000, 37000, 41000, 45000]
}
```

```python
df = pd.DataFrame(data)
df
```

**Step 3 : Inspect the Dataset**

```python
df.head()
```

**Step 4 : Generate Descriptive Statistics**

```python
df.describe()
```

**Step 5 : Calculate Mean**

```python
# Calculate the average number of units sold per month.

average_units = df['Units_Sold'].mean()
print("Average Units Sold : ",average_units)
```

**Step 6 : Calculate Median**

```python
# Calculate the middle value of the units column

median_units = df['Units_Sold'].median()
print("Median units sold : ",median_units)
```

**Step 7: Calculate the Maximum and Minimum**

```python
# Find the maximum unit sold in a month

print("Maximum Units sold in a month", df['Units_Sold'].max())

# Find the minimum unit sold in a month

print("Maximum Units sold in a month", df['Units_Sold'].min())
```

**Step 8 : Calculate Standard Deviation**

```python
# Calculates the standard deviation of Units Sold
std_units = df['Units_Sold'].std()

print("Standard Deviation of Units Sold:", std_units)
```

**Step 9 : Find the Month with highest sale**

```python
# Find the index of the row having the maximum revenue

highest_sales_index = df['Revenue'].idxmax()

#Retrieve the month corresponding to that index

highest_sales_month = df.loc[highest_sales_index,'Months']

# Retrieve the maximum revenue

highest_revenue = df.loc[highest_sales_index, "Revenue"]

#Display the Result

print("Highest Sales Month : ",highest_sales_month)
print("Highest Revenue : ",highest_revenue)
```

**Step 10 : Find the Month with lowest sale**

```python
# Find the index of the row having the maximum revenue

lowest_sales_index = df['Revenue'].idxmin()

#Retrieve the month corresponding to that index

lowest_sales_month = df.loc[lowest_sales_index,'Months']

# Retrieve the maximum revenue

lowest_revenue = df.loc[lowest_sales_index, "Revenue"]

#Display the Result

print("Lowest Sales Month : ",lowest_sales_month)
print("Lowest Revenue : ",lowest_revenue)
```

**Step 11 : Calculate Total Revenue**

```python
# Calculate the total revenue generated during the year

total_revenue = df["Revenue"].sum()
print("Total Revenue :  ",total_revenue)
```

**Step 12 : Calculate Total Profit**

```python
# Calculate the total profit

total_profit = df['Profit'].sum()
print("Display Profit : ",total_profit)
```

**Step 13 : Identify the Sales Trend**

```python
from matplotlib.typing import LineStyleType
# Create a line chart to visualize monthly revenue
plt.figure(figsize=(10,5))
#Plot Month on the X-axis and Revenue on the Y-axis.
plt.plot(df['Months'],df['Revenue'],marker='o', linestyle="--")
# Add a title to the graph
plt.title("Sales Trends")
#Add a label to the X-axis
plt.xlabel("Months")
# Add a label to Y-axis
plt.ylabel("Revenue")
plt.xticks(rotation=45)
# Display the figure
plt.show()
```

**Step 14 : Analyize the Profit Trend**

```python
# Create a line chart to visualize monthly Profit
plt.figure(figsize=(10,5))
#Plot Month on the X-axis and Profit on the Y-axis.
plt.plot(df['Months'],df['Profit'],marker='o', linestyle="--")
# Add a title to the graph
plt.title("Sales Trends")
#Add a label to the X-axis
plt.xlabel("Months")
# Add a label to Y-axis
plt.ylabel("Profit")
plt.xticks(rotation=45)
# Display the figure
plt.show()
```

**Step 15 : Compare Revenue and Profit**

```python
# Create a line chart to compare Revenue and Profit
plt.figure(figsize=(12,8))
# Plot Revenue
plt.plot(df['Months'],df['Revenue'],marker='o', linestyle="--",label="Revenue")
# Plot Profit
plt.plot(df['Months'],df['Profit'],marker='o', linestyle="--",label="Profit")
# Add label to X-axis
plt.xlabel("Months")
plt.xticks(rotation=45)
# Add label to Y-axis
plt.ylabel("Revenue/Profit")
# Add a Title for graph
plt.title("Revenue vs Profit")
# Add legend for identifying line
plt.legend()
# Display the graph
plt.show()
```

**Step 16 : Identify the growth in sales**

```python
# Calculate the  percentage of change in revenue
# from one month to next month
df['Revenue_Growth'] = df['Revenue'].pct_change() * 100
df
```

**Step 17 : Identify the month with increasing Revenue**

```python
# Filter the dataset to display only those months
# where the revenue growth is positive

increasing_sales = df[df['Revenue_Growth']>0]

increasing_sales
```
