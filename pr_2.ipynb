step 1 : Import pandas as pd

```python
# import pandas and create dataframe from data
import pandas as pd

data={
    "ID": [ 101, 102, 103, 104, 105, 106, 107,108],
    "Name": ["Rahul", "priya", "Amit", None,"Riya", "karan", "Jay", "soham"],
    "Age": [20, 21,None, 19, 20, 22, 24, 25],
    "City": ["surat", "Navsari", "Valsad", "Rajkot", "Surat", "Gandhinagar", "Ahmedabad",None],
    "Marks":[85, None, 78, 88, 81, 84, 90, 92]
}
ys = pd.DataFrame(data)
ys
```

Stpe 4 : Check Missing Values

```python
# count the missing values from dataset.
ys.isnull().sum()
```

```python
# Diaplay a boolean dataframe.
ys.isnull().sum().sum()
```

```python
# Calculate number of missing values in the dataframe. mean() calculate the average marks. fillna() replaces missing values.
ys["Marks"] = ys["Marks"].fillna(ys["Marks"].mean())

ys
```

Fill Missing Age

```python
# Replace missing values in age column
# using the average age of all students.

ys["Age"] = ys["Age"].fillna(ys["Age"].mean()).astype(int)

ys
```

Fill Missing Name

```python
# Replace missing values in the age column
# using the average age of all students.

ys["Name"] = ys["Name"].fillna("unknown")

ys
```

Step 8 : Rename a column

```python
#  Rename the column "Marks" to "Score". inplace True permanently upadates the Dataframe
ys.rename(columns={"Marks":"score"}, inplace=True)

ys
```

Step 9 : Change Data Type

```python
# convert age column from float value to integer.
ys['score'] = ys['score'].astype(int)
ys.dtypes
```

Step 10 : Add final score column by adding 5 marks

```python
# create a new column and add 5 marks to the new column"score"
ys["final_score"] = ys["score"] + 5
ys
```

Step 11 : convert text to uppercase

```python
# convert a name from case to uppercase.
ys["Name"]=ys["Name"].str.upper()
ys
```

Step 12 : Convert text to lowercase

```python
# Convert city from case to lowercase
ys["City"] = ys["City"].str.lower()
ys
```

Step 13 : Remove Extra Space

```python
# Remove extra space
ys["City"] = ys["City"].str.strip()
ys
```

Step 14 : Sort the dataset

```python
# Based on "Final_Score"sort a data.
ys.sort_values(by="final_score", ascending=False)
```

Step 15 : Save the cleaned dataset

```python
# Save the clean dataset
ys.to_csv("cleaned_student.CSV",index=False)
print("Dataset create successfuly")
```
