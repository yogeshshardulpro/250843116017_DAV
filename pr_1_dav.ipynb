Practical 1 | Study the Data Analytics lifecycle and perform data acquisition from CSV, Excel, and JSON files using Pandas

Step 1 : Import Pandas

```python
# Import pandas library.
import pandas as pd

data = {
    "ID": [ 101, 102, 103, 104, 105, 106, 107],
    "Name": ["Rahul", "priya", "Amit", "Riya", "karan", "Krishna", "Harry"],
    "Age": [20, 21, 19, 20, 22, 24, 28],
    "City": ["surat", "Navsari", "Valsad", "Bharuch", "Surat", "Gandhinagar", "Ahmedabad"],
    "Marks":[85, 92, 78, 88, 81, 84, 90]
   }


```

Crate a dataframe

```python
# create a DataFrame from given data
# Display the DataFrame
df = pd.DataFrame(data)
df
```

Step 4 : Save Data as CSV File

```python
# Export the DataFrame to a csv file
df.to_csv("data.csv", index=False)

print("csv file created")
```

Step 5 : Save Data as Excel File

```python
# Export the DataFrame to an Excel file
df.to_excel("students_info.xlsx", index=False)
print("excel file created")
```

Step 6 : Save Data as JSON file

```python
# Export the DataFrame to a JSON  file in records format
df.to_json("students.json", orient="records", indent=5)

print("json file created")
```
