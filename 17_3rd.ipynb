```python
# Import the required modules such as pandas and Numpy
import pandas as pd
import numpy as np
```

**Step 1: Create a Dataframe/Records**

```python
# All fields and Values
data = {
    "Emp_ID" : [101, 102, 103, 104, 105, 106, 107, 108],
    "Name" : ["Rahul","Priya","Amit","Riya","Karan","Neha","Vivek","Pooja"],
    "Department" : ["IT","HR","IT","Finance","HR","Finance","IT","HR"],
    "City" : ["Surat","Navsari","Surat","Valsad","Surat","Navsari","Valsad","Surat"],
    "Salary":[45000, 38000, 52000, 48000, 40000, 51000, 60000, 42000],
    "Experience":[2,1,5,4,2,6,8,3]
}
```

```python
# Create DataFrame using pandas method
df = pd.DataFrame(data);
```

```python
# Display DataFrame
df
```

**Step 2: Filter Data**

2.1 Display employee from IT Department

```python
# Select only employees working in IT department
# The condition return only those row where Department is equal to "IT"
df[df["Department"] == "IT"]
```

2.2 Display Employees with salary greater than 20000

```python
# Find Employees whose salary is greater than 20000 using greater than operator
df[df['Salary'] > 20000]
```

2.3 Display Employee from Surat

```python
# Find Employees whose city is equal to Surat using greater than operator
df[df['City'] == "Surat"]
```

2.4 Apply Multiple Condition

```python
# Display employees from IT department having salary greater than 50000
df[(df['Department'] == "IT") & (df['Salary'] > 20000)]
```

**Step 3: Sorting Data**

3.1 Sort Salary in Ascending Order

```python
# Sort Salary from lowest to Highest
df.sort_values(by="Salary")
```

3.2 Sort Salary in descending order

```python
# Sort Salary from highest to lowest
df.sort_values(by="Salary", ascending=False)
```

3.3 Sort Department Wise Salary

```python
# First Sort by department
# Within each department, sort salaries in descending order
df.sort_values(by=["Department", "Salary"], ascending=[True,False])
```

**Step 4: Perform Aggregation**

4.1 Calculate Total Salary

```python
# Calculate the Total Salary By Sum
df["Salary"].sum()
```

4.2 Calculate the avg salary

```python
# Calculate the average salary by mean
df["Salary"].mean()
```

4.3 Find Highest Salary

```python
# Find the highest salary by max
df['Salary'].max()
```

4.4 Find Lowest Salary

```python
# Find the lowest salary by min
df["Salary"].min()
```

4.5 Count Employees

```python
# Find the total employess using count method
df['Emp_ID'].count()
```

**Step 5:- Perform groupBy Operation**

5.1 Group Employee Based on department and Claculate the avg salary of each department

```python
# Group employees by Department
# Calculate the average salary of each department
df.groupby('Department')['Salary'].mean()
```

5.2 Calculate the total salary paid by department

```python
# Group employees by Department
# Calculate the total salary of each department
df.groupby('Department')['Salary'].sum()
```

5.3 Display the highest salary in each department

```python
# Group employees by Department
# Calculate the highest salary of each department
df.groupby('Department')['Salary'].max()
```

5.4 Count Employees Department Wise

```python
# Group employees by Department
# Count Employess of each department
df.groupby('Department')['Salary'].count()
```

5.5 Calculate the average salary of employees in each city

```python
# Group employees by City
# Calculate the average salary of each city
df.groupby('City')['Salary'].mean()
```

**Step 6 : Multiple Aggregation Function**

```python
# Apply multiple statistical functions to the salary column for each department
df.groupby('Department')['Salary'].agg(['count', 'sum' ,'mean', 'min', 'max'])
```

**Step 7 : save the result**

```python
# Save the dataset into CSV file

df.to_csv("Employee_Data.csv", index=False)

print("Dataset Saved Successfully")
```
