### **Practical no :- 6 | Create basic data visualizations (Line, Bar, Histogram, Scatter, Pie).**

**Step 1 :- Libaries**

```python
# Import Necessary Library
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Suppress all warnings for cleaner output
import warnings
warnings.filterwarnings('ignore')
```

**Step 2 :- Create a Dataset**

```python
# Define sales data containing Month, Product, Region, Sales, and Units_Sold
data = {
    "Month" : ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'],
    "Product" : ['Laptop','Mobile','Tablet','Laptop','Mobile','Tablet','Laptop','Mobile','Tablet','Laptop','Mobile','Tablet'],
    "Region" : ["North",'South','East','West',"North",'South','East','West',"North",'South','East','West'],
    "Sales" : [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
    "Units_Sold" : [12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40]
}
```

```python
# Create a Pandas DataFrame from the defined sales data
df = pd.DataFrame(data)
```

**Step 3 : Create a line chart**

```python
# Visualize monthly sales using a line plot
# Set figure size for better readability
plt.figure(figsize=(10.6, 6))
# Plot 'Sales' against 'Month' with markers at each data point
plt.plot(df['Month'], df['Sales'], marker='o')
# Add a grid for easier data interpretation
plt.grid(True, which='both', linestyle=':', linewidth=0.5, color='black')
# Set the title of the plot
plt.title('Monthiwse company sales')
# Label the x-axis
plt.xlabel('Month')
# Label the y-axis
plt.ylabel('Sales')
# Display the plot
plt.show()
```

**Step 4 :-  Create a bar chart**

```python
# Calculate total sales for each product
labels = df.groupby('Product')['Sales'].sum()
```

```python
# Create a bar chart to visualize sales per product
# Set figure size for better readability
plt.figure(figsize=(10,6))
# Create the bar plot with 'Product' on the x-axis and 'Sales' on the y-axis
bars = plt.bar(df['Product'],df['Sales'])
# Set the title of the plot
plt.title('Product wise sales')
# Label the x-axis
plt.xlabel('Product')
# Label the y-axis
plt.ylabel('Sales')
# Add the calculated total sales as labels on top of each bar
plt.bar_label(container=bars, labels=labels)
# Display the plot
plt.show()
```

**Step 5 :- Create a histogram**

```python
# Visualize the distribution of individual sales values using a histogram with KDE
# Set figure size for better readability
plt.figure(figsize=(10,6))
# Create a histogram of 'Sales' with a Kernel Density Estimate (KDE) overlay
sns.histplot(x = df['Sales'],kde=True)
# Set the title of the plot
plt.title('Sales Distribution')
# Label the x-axis
plt.xlabel('Sales')
# Label the y-axis
plt.ylabel('Frequency')
# Display the plot
plt.show()
```

**Step 6 :- Create a scatter plot**

```python
# Create a scatter plot to examine the relationship between units sold and total sales
# Set figure size for better readability
plt.figure(figsize=(10,6))
# Create the scatter plot with 'Units_Sold' on the x-axis and 'Sales' on the y-axis
sns.scatterplot(x = df['Units_Sold'], y = df['Sales'])
# Add a grid for easier data interpretation
plt.grid(True, which='both', linestyle=':', linewidth=0.5, color='black')
# Set the title of the plot
plt.title("Unit Sold vs Sales")
# Label the x-axis
plt.xlabel("Units Sold")
# Label the y-axis
plt.ylabel("Sales")
# Display the plot
plt.show()
```

**Step 7 :- Create a pie chart**

```python
# Calculate the total sales for each region
data = df.groupby('Region')['Sales'].sum()
```

```python
# Create a pie chart to show the percentage of total sales from each region
# Set figure size for better readability
plt.figure(figsize=(10,5))
# Create the pie chart using the grouped 'data', region names as labels, and show percentages
plt.pie(data, labels= df['Region'].unique(), autopct='%1.1f%%', startangle=140)
# Add a legend to identify regions
plt.legend()
# Display the plot
plt.show()
```
